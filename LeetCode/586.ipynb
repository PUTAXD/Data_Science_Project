```python
import pandas as pd
```

```python
data = [[1, 1], [2, 2], [3, 3], [4, 3]]
orders = pd.DataFrame(data, columns=['order_number', 'customer_number']).astype({'order_number':'Int64', 'customer_number':'Int64'})
```

```python
orders
```

```python
orders = orders.groupby('customer_number')['order_number'].sum().reset_index()
```

```python
orders.sort_values('order_number', ascending=False).head(1)
```
