```python
import pandas as pd

data = [[1, 'a@b.com'], [2, 'c@d.com'], [3, 'a@b.com'], [4, 'a@b.com']]
person = pd.DataFrame(data, columns=['id', 'email']).astype({'id':'Int64', 'email':'object'})
person
```

```python


person["duplicat"] = person["email"].duplicated()
result = person[person["duplicat"] == True]
result.drop(columns=["duplicat"],axis=1).drop(columns=["id"],axis=1).rename(columns={"email":"Email"})
```

```python
duplicate_rows = person.loc[person.duplicated('email') == True]
result = duplicate_rows[['email']].drop_duplicates()
result    
```
