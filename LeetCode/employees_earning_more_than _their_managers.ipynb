```python
import pandas as pd
```

```python
data = [[1, 'Joe', 70000, 3], [2, 'Henry', 80000, 4], [3, 'Sam', 60000, None], [4, 'Max', 90000, None]]
employee = pd.DataFrame(data, columns=['id', 'name', 'salary', 'managerId']).astype({'id':'Int64', 'name':'object', 'salary':'Int64', 'managerId':'Int64'})

employee
```

```python
employee_managers = employee.merge(employee,how="inner", left_on="managerId", right_on="id")
salary = employee_managers[["name_x","salary_x","salary_y"]]

salary
```

```python
result = salary[salary["salary_x"] > salary["salary_y"]]
result
```

```python
result = result.rename(columns={"name_x" : "Employee"}).drop("salary_y",axis=1).drop("salary_x",axis=1)
result
```
